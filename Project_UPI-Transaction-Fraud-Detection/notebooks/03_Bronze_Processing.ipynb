# 03 - Bronze Processing
Loads Landing data as-is into a Delta table, adding ingestion metadata only.
No cleaning happens here — Bronze preserves the raw, messy data + when/where it was ingested.

```python
from pyspark.sql.functions import current_timestamp, input_file_name, to_date

LANDING_PATH = "/Volumes/sentinel_catalog/sentinel_schema/raw_volume/landing"
BRONZE_TABLE = "sentinel_catalog.sentinel_schema.bronze_transactions"

spark.sql("CREATE CATALOG IF NOT EXISTS sentinel_catalog")
spark.sql("CREATE SCHEMA IF NOT EXISTS sentinel_catalog.sentinel_schema")

landing_df = spark.read.json(LANDING_PATH)
```

```python
from pyspark.sql.functions import current_timestamp, to_date

bronze_df = (
    landing_df
    .withColumn("ingestion_timestamp", current_timestamp())
    .withColumn("ingestion_date", to_date(current_timestamp()))
)

(
    bronze_df.write
    .format("delta")
    .mode("append")
    .partitionBy("ingestion_date")
    .option("mergeSchema", "true")
    .saveAsTable(BRONZE_TABLE)
)

print(f"Wrote {bronze_df.count()} rows to {BRONZE_TABLE}")
```
