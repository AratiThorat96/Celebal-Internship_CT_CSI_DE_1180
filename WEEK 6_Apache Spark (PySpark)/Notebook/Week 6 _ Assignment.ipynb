# Week 6 - Apache Spark Assignment

## Objective
The objective of this assignment is to understand Apache Spark architecture and perform data processing using PySpark DataFrames. This includes reading data, applying transformations, filtering records, handling schema, and comparing file formats such as CSV and Parquet.

## Step 1: Import Required Libraries

In this step, we import the SparkSession class, which is the entry point for working with Apache Spark using PySpark.

```python
# Import SparkSession
from pyspark.sql import SparkSession
```

## Step 2: Create Spark Session

SparkSession is the starting point of every Spark application.
It allows us to create DataFrames and interact with the Spark engine.

```python
# Create Spark Session
spark = SparkSession.builder \
    .appName("Week6_Assignment") \
    .getOrCreate()

print("Spark Version:", spark.version)
```

## Step 3: Read the CSV Dataset

The dataset is stored in the Databricks Volume. We use the header option to read the first row as column names and inferSchema to automatically detect data types.

```python
# Read CSV file
df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/workspace/default/data_week6/source.csv")
```

## Step 4: Display the Dataset

The show() function displays the first few records of the DataFrame.

```python
# Display first 5 rows
df.show(5)
```

## Step 5: Print Schema

The schema shows the column names and their respective data types.
Understanding the schema is important before applying transformations.

```python
# Print schema
df.printSchema()
```

## Step 6: Count Total Records

The count() action returns the total number of rows in the dataset.

```python
# Count rows
print("Total Records:", df.count())
```

## Step 7: Check Null Values

Before processing data, it is a good practice to identify missing values.

```python
from pyspark.sql.functions import col, sum

null_df = df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
])

null_df.show()
```

## Step 8: Remove Duplicate Records

Duplicate records can affect the accuracy of analysis.

```python
# Remove duplicate rows
df = df.dropDuplicates()

print("Duplicates removed successfully.")
```

## Step 9: Filter Records

This example demonstrates filtering records using a condition.

```python
# Example filter
filtered_df = df.filter(df.category == "Electronics")

filtered_df.show()
```

## Step 10: Rename a Column

```python
df = df.withColumnRenamed("old_name", "new_name")
```

## Step 11: Change Data Type

```python
from pyspark.sql.functions import col

df = df.withColumn(
    "price",
    col("price").cast("double")
)
```

## Step 12: Add a New Column

```python
df = df.withColumn(
    "final_price",
    col("base_price") * 1.18
)
```

## Step 13: Save Data as Parquet

Parquet is a columnar storage format that provides better compression and faster query performance.

```python
df.write.mode("overwrite").parquet(
    "/Volumes/workspace/default/data_week6/output_parquet"
)
```

## Step 14: Save Data as CSV

```python
df.write.mode("overwrite") \
.option("header","true") \
.csv("/Volumes/workspace/default/data_week6/output_csv")
```
